## 3-4_Kernel_PCA_for_nonlinear_dimensionality_reduction

```python
## 1. Load the Iris dataset and reduce it to two dimensions using Kernel PCA.

from sklearn import datasets, decomposition
iris = datasets.load_iris()
iris_X = iris.data


kpca = decomposition.KernelPCA(kernel='cosine', n_components=2)
iris_X_prime = kpca.fit_transform(iris_X)
```
