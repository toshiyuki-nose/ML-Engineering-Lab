## 3-5_Using_truncated_SVD_to_reduce_dimensionality

```python
## 1. Load the Iris dataset and set the explanatory and target variables.

from sklearn.datasets import load_iris
iris = load_iris()
iris_X = iris.data
y = iris.target
```

```python
## 2. Reduce the Iris dataset to two dimensions using Truncated SVD.

from sklearn.decomposition import TruncatedSVD
svd = TruncatedSVD(2)
iris_transformed = svd.fit_transform(iris_X)
```

```python
## 3. Display the explained variance ratio of the two components.

svd.explained_variance_ratio_
```

```python
## 4. Create a small matrix for demonstrating singular value decomposition.

from scipy.linalg import svd
import numpy as np
D = np.array([[1, 2], [1, 3], [1, 4]])
D
```

```python
## 5. Perform singular value decomposition and display the shapes of the resulting matrices.

U, S, V = svd(D, full_matrices=False)
U.shape, S.shape, V.shape
```

```python
## 6. Reconstruct the original matrix from the SVD components.

np.dot(U.dot(np.diag(S)), V)
```
