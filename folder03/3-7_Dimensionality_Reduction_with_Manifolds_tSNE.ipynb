## 3-7_Dimensionality_Reduction_with_Manifolds_tSNE

```python
## 1. Load the Iris dataset and set the explanatory and target variables.

from sklearn.datasets import load_iris
iris = load_iris()
iris_X = iris.data
y = iris.target
```

```python
## 2. Import PCA and manifold learning methods and prepare Matplotlib for visualization.

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE, MDS, Isomap

#Load visualization library
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
## 3. Reduce the Iris dataset to two dimensions using PCA and visualize the result.

iris_pca = PCA(n_components=2).fit_transform(iris_X)

plt.figure(figsize=(10, 7))
plt.scatter(iris_pca[:, 0], iris_pca[:, 1], c=y)
plt.title("PCA")
```

```python
## 4. Reduce the Iris dataset to two dimensions using t-SNE and visualize the result.

iris_tsne = TSNE(learning_rate=200).fit_transform(iris_X)

plt.figure(figsize=(10, 7))
plt.scatter(iris_tsne[:, 0], iris_tsne[:, 1], c=y)
plt.title("TSNE")
```

```python
## 5. Reduce the Iris dataset to two dimensions using MDS and visualize the result.

iris_mds = MDS(n_components=2).fit_transform(iris_X)

plt.figure(figsize=(10, 7))
plt.scatter(iris_mds[:, 0], iris_mds[:, 1], c=y)
plt.title("MDS")
```

```python
## 6. Reduce the Iris dataset to two dimensions using Isomap and visualize the result.

iris_isomap = Isomap(n_components=2).fit_transform(iris_X)

plt.figure(figsize=(10, 7))
plt.scatter(iris_isomap[:, 0], iris_isomap[:, 1], c=y)
plt.title("Isomap")
```

```python
## 7. Apply t-SNE to the PCA-transformed Iris dataset and visualize the result.

iris_pca_then_tsne = TSNE(learning_rate=200).fit_transform(iris_pca)
plt.figure(figsize=(10, 7))
plt.scatter(iris_pca_then_tsne[:, 0], iris_pca_then_tsne[:, 1], c=y)
plt.title("PCA followed by TSNE")
```
