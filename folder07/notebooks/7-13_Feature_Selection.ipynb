## 7-13_Feature_Selection

```python
## 1. Generate a regression dataset with 10,000 features.

from sklearn import datasets
X, y = datasets.make_regression(1000, 10000)
```

```python
## 2. Calculate F-statistics and p-values for each feature.

from sklearn import feature_selection
f, p = feature_selection.f_regression(X, y)
```

```python
## 3. Display the first five F-statistics.

f[:5]
```

```python
## 4. Display the first five p-values.

p[:5]
```

```python
## 5. Select features with p-values below 0.05.

import numpy as np
idx = np.arange(0, X.shape[1])
features_to_keep = idx[p < .05]
len(features_to_keep)
```

```python
## 6. Remove low-variance features using VarianceThreshold.

var_threshold = feature_selection.VarianceThreshold(np.median(np.var(X, axis=1)))
var_threshold.fit_transform(X).shape
```

```python
## 7. Generate a new regression dataset and calculate F-statistics and p-values.

X, y = datasets.make_regression(10000, 20)
f, p = feature_selection.f_regression(X, y)
```

```python
## 8. Visualize the p-values of the 20 features.

%matplotlib inline
from matplotlib import pyplot as plt
f, ax = plt.subplots(figsize=(7, 5))
ax.bar(np.arange(20), p, color='k')
ax.set_title("Feature p values")
```
