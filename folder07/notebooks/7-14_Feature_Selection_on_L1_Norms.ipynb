## 7-14_Feature_Selection_on_L1_Norms

```python
## 1. Load the diabetes dataset.

import sklearn.datasets as ds
diabetes = ds.load_diabetes()

X = diabetes.data
y = diabetes.target
```

```python
## 2. Initialize a linear regression model.

from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
## 3. Import evaluation metrics and cross-validation utilities.

from sklearn.metrics import make_scorer, mean_squared_error
from sklearn.model_selection import cross_val_score,ShuffleSplit
```

```python
## 4. Evaluate linear regression before feature selection.

shuff = ShuffleSplit(n_splits=10, test_size=0.25, random_state=0)
score_before = cross_val_score(lr,X,y,cv=shuff,scoring=make_scorer(mean_squared_error,greater_is_better=False)).mean()

score_before
```

```python
## 5. Fit LassoCV and inspect the feature coefficients.

from sklearn.linear_model import LassoCV

lasso_cv = LassoCV()
lasso_cv.fit(X,y)
lasso_cv.coef_
```

```python
## 6. Select features with non-zero Lasso coefficients.

import numpy as np
columns = np.arange(X.shape[1])[lasso_cv.coef_ != 0]
columns
```

```python
## 7. Evaluate linear regression after Lasso feature selection.

score_afterwards = cross_val_score(lr,X[:,columns],y,cv=shuff, scoring=make_scorer(mean_squared_error,greater_is_better=False)).mean()
score_afterwards
```

```python
## 8. Generate synthetic regression data and evaluate the baseline model.

X, y = ds.make_regression(noise=5)
shuff = ShuffleSplit(n_splits=10, test_size=0.25, random_state=0)

score_before = cross_val_score(lr,X,y,cv=shuff, scoring=make_scorer(mean_squared_error,greater_is_better=False)).mean()
```

```python
## 9. Compare regression performance before and after Lasso feature selection.

lasso_cv = LassoCV()
lasso_cv.fit(X,y)

columns = np.arange(X.shape[1])[lasso_cv.coef_ != 0]
score_afterwards = cross_val_score(lr,X[:,columns],y,cv=shuff, scoring=make_scorer(mean_squared_error,greater_is_better=False)).mean()
print("Score before:",score_before)
print("Score after: ",score_afterwards)
```
