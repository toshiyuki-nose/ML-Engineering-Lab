## 5-3_Viewing_the_Pima_Indians_Diabetes_Dataset_with_Pandas

```python
## 1. Load the Pima Indians Diabetes Dataset and display the first five rows.

import pandas as pd

data_web_address = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.csv"
column_names = ['pregnancy_x', 
                'plasma_con', 
                'blood_pressure', 
                'skin_mm', 
                'insulin', 
                'bmi', 
                'pedigree_func', 
                'age', 
                'target']

feature_names = column_names[:-1]
all_data = pd.read_csv(data_web_address , names=column_names)
all_data.head()
```

```python
## 2. Display descriptive statistics for the dataset.

all_data.describe()
```

```python
## 3. Visualize the distribution of pregnancy counts with a histogram.

%matplotlib inline

#The default is bins=10 which is hard to read in the visualization.
all_data.pregnancy_x.hist(bins=50)
```

```python
## 4. Visualize histograms for all numerical columns.

all_data.hist(figsize=(15,9),bins=50)
```

```python
## 5. Count the occurrences of each target class.

all_data.target.value_counts()
```
