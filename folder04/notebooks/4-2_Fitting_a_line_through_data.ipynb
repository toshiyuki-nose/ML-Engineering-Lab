## 4-2_Fitting_a_line_through_data

```python
## 1. Load the housing dataset and fit a linear regression model.

from sklearn import datasets
boston = datasets.fetch_california_housing()

from sklearn.linear_model import LinearRegression
lr = LinearRegression()

lr.fit(boston.data, boston.target)
predictions = lr.predict(boston.data)
```

```python
## 2. Visualize the distribution of residuals using a histogram.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline

pd.Series(boston.target - predictions).hist(bins=50)
```

```python
## 3. Pair feature names with their corresponding linear regression coefficients.

zip(boston.feature_names, lr.coef_)
```

```python
## 4. Display the coefficients of the fitted linear regression model.

lr.coef_
```

```python
## 5. Display the intercept of the fitted linear regression model.

lr.intercept_
```

```python
## 6. List the attributes and methods of the fitted linear regression model.

dir(lr)
```

```python
## 7. Fit a linear regression model with feature normalization.

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

lr2 = make_pipeline(StandardScaler(), LinearRegression())
lr2.fit(boston.data, boston.target)
```

```python
## 8. Predict housing prices using the fitted regression pipeline.

predictions2 = lr2.predict(boston.data)
```
