## 6-3_Optimizing_the_number_of_centroids

```python
## 1. Generate synthetic data with three clusters.

from sklearn.datasets import make_blobs
import numpy as np
blobs, classes = make_blobs(500, centers=3)
```

```python
## 2. Train the K-Means model with three clusters.

from sklearn.cluster import KMeans
kmean = KMeans(n_clusters=3)
kmean.fit(blobs)
```

```python
## 3. Calculate silhouette scores for individual samples.

from sklearn import metrics
silhouette_samples = metrics.silhouette_samples(blobs, kmean.labels_)
np.column_stack((classes[:5], silhouette_samples[:5]))
```

```python
## 4. Calculate the mean silhouette score.

silhouette_samples.mean()
```

```python
## 5. Calculate the overall silhouette score directly.

metrics.silhouette_score(blobs, kmean.labels_)
```

```python
## 6. Compare silhouette scores across different numbers of clusters.

import matplotlib.pyplot as plt
%matplotlib inline

blobs, classes = make_blobs(500, centers=10)
silhouette_avgs = []
for k in range(2, 60):
     kmean = KMeans(n_clusters=k).fit(blobs)
     silhouette_avgs.append(metrics.silhouette_score(blobs, kmean.labels_))

f, ax = plt.subplots(figsize=(7, 5))
ax.plot(silhouette_avgs)
```
