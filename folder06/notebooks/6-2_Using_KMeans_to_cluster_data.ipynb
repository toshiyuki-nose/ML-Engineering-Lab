## 6-2_Using_KMeans_to_cluster_data

```python
## 1. Generate synthetic data for clustering.

import numpy as np
import pandas as pd

from sklearn.datasets import make_blobs
blobs, classes = make_blobs(500, centers=3)
```

```python
## 2. Import Matplotlib for data visualization.

import matplotlib.pyplot as plt
%matplotlib inline
```

```python
## 3. Visualize the generated data as a scatter plot.

f, ax = plt.subplots(figsize=(7.5, 7.5))
rgb = np.array(['r', 'g', 'b'])
ax.scatter(blobs[:, 0], blobs[:, 1], color=rgb[classes])
ax.set_title("Blobs")
```

```python
## 4. Train the K-Means clustering model.

from sklearn.cluster import KMeans
kmean = KMeans(n_clusters=3)

kmean.fit(blobs)
```

```python
## 5. Inspect the coordinates of the cluster centers.

kmean.cluster_centers_
```

```python
## 6. Visualize the cluster centers on the scatter plot.

f, ax = plt.subplots(figsize=(7.5, 7.5))
ax.scatter(blobs[:, 0], blobs[:, 1], color=rgb[classes])
ax.scatter(kmean.cluster_centers_[:, 0],kmean.cluster_centers_[:, 1], marker='*', s=250,color='black', label='Centers')
ax.set_title("Blobs")
ax.legend(loc='best')
```

```python
## 7. Inspect the first five predicted cluster labels.

kmean.labels_[:5]
```

```python
## 8. Inspect the first five original class labels.

classes[:5]
```

```python
## 9. Calculate distances from data points to cluster centers.

kmean.transform(blobs)[:5]
```
