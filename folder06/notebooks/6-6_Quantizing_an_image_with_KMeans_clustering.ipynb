## 6-6_Quantizing_an_image_with_KMeans_clustering

```python
## 1. Load and display the original image.

%matplotlib inline
import matplotlib.pyplot as plt
from scipy import ndimage
img = plt.imread("headshot.jpg")
plt.figure(figsize = (10,7))
plt.imshow(img)
```

```python
## 2. Inspect the dimensions of the image.

img.shape
```

```python
## 3. Reshape the image into a two-dimensional array.

x, y, z = img.shape
long_img = img.reshape(x*y, z)
long_img.shape
```

```python
## 4. Apply K-Means clustering to the image pixels.

from sklearn import cluster
k_means = cluster.KMeans(n_clusters=5)
k_means.fit(long_img)
centers = k_means.cluster_centers_
centers
```

```python
## 5. Inspect the cluster labels assigned to each pixel.

labels = k_means.labels_
labels
```

```python
## 6. Reconstruct and display the quantized image.

plt.figure(figsize = (10,7))
plt.imshow(centers[labels].reshape(x, y, z).astype("uint8"))
```
