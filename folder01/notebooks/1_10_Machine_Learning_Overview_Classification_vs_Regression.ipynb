## 1_10_Machine_Learning_Overview_Classification_vs_Regression

```python
## 1. Load the iris dataset from the datasets module.
import numpy as np
import pandas as pd
from sklearn import datasets

iris = datasets.load_iris()
```

```python
## 2. Consider only the values ​​0 (Setosa) and 1 (Versicolor) for the target variable.
## Exclude the target variable value of 2 using the boolean array `iris.target < 2`.
## By enclosing this array in square brackets `[]`, use it as a filter when defining the feature dataset `X` and the target dataset `y`.

X = iris.data[iris.target < 2]
y = iris.target[iris.target < 2]
```

```python
## 3. Import the train_test_split function and apply it to these datasets.
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state= 7)
```

```python
## 4. Import SVC, fit it to the data, and then calculate the accuracy using cross-validation.

from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score

svc_clf = SVC(kernel = 'linear').fit(X_train, y_train)
svc_scores = cross_val_score(svc_clf, X_train, y_train, cv=4)
```

```python
## 5. Calculate the average accuracy rate, as before.
## In scikit-learn 0.20, the average accuracy is 1.0.
svc_scores.mean()
```

```python
## 6. Import Support Vector Regression (SVR) from sklearn.svm.

from sklearn.svm import SVR
```

```python
## 7. Write the syntax required to fit the SVR model. The syntax is almost identical to that of SVC; you simply replace the keyword 'c' with 'r'.

svr_clf = SVR(kernel = 'linear').fit(X_train, y_train)
```

### Create a score function.
To create a score function, the following two elements are required:
- A function that compares `y_test` (ground truth) with `y_pred` (predicted values)
- A determination of whether a higher score is better or worse

```python
## 8. First, import the make_scorer function.

from sklearn.metrics import make_scorer
```

```python
## 9. Use the following sample score function.

#Only works for this iris example with targets 0 and 1
def for_scorer(y_test, orig_y_pred):
    y_pred = np.rint(orig_y_pred).astype(int)   #rounds prediction to the nearest integer
    return accuracy_score(y_test, y_pred)
```

```python
## 10. Determine whether a higher score is better. In this situation, a higher accuracy rate is preferable, so it is determined that a higher score is better.

svr_to_class_scorer = make_scorer(for_scorer, greater_is_better=True) 
```

```python
## 11. Finally, perform cross-validation using the new parameter (Scoring).

svr_scores = cross_val_score(svr_clf, X_train, y_train, cv=4, scoring = svr_to_class_scorer)
```

```python
## 12. Calculate the average value.

svr_scores.mean()
```
