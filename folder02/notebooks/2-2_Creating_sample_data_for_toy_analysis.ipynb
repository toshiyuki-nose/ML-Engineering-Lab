```python
## 3. First, let's look at regression.
## By default, the following code generates a tuple containing a 100x100 matrix.

reg_data = d.make_regression()
```

```python
## 4. For example, to generate a 1000×10 matrix with five features possessing information gain (i.e., highly correlated with the target variable's values), a bias term of 1.0, and two target variable values, execute the following command.


complex_reg_data = d.make_regression(
    n_samples=1000,
    n_features=10,
    n_informative=5,
    n_targets=2,
    bias=1.0
)

complex_reg_data[0].shape
```
