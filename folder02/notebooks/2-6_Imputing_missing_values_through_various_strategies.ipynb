## 2-6_Imputing_missing_values_through_various_strategies

```python
## 1. Load the Iris dataset and randomly introduce missing values.

from sklearn import datasets
import numpy as np

iris = datasets.load_iris()
iris_X = iris.data

masking_array = np.random.binomial(
    1,
    0.25,
    iris_X.shape
).astype(bool)

iris_X[masking_array] = np.nan
```

```python
## 2. Display the first five rows of the missing-value mask.

masking_array[:5]
```

```python
## 3. Display the first five rows of the Iris dataset containing missing values.

iris_X[:5]
```

```python
## 4. Impute missing values using the mean of each feature.

from sklearn.impute import SimpleImputer

impute = SimpleImputer(strategy="mean")

iris_X_prime = impute.fit_transform(iris_X)

iris_X_prime[:5]
```

```python
## 5. Display a value that was imputed with the mean.

iris_X_prime[0, 0]
```

```python
## 6. Display the original value before imputation for comparison.

iris_X[0, 0]
```

```python
## 7. Impute missing values using the median of each feature.

from sklearn.impute import SimpleImputer

impute = SimpleImputer(strategy="median")

iris_X_prime = impute.fit_transform(iris_X)

iris_X_prime[:5]
```

```python
## 8. Replace all missing values with a constant value of -1.

iris_X[np.isnan(iris_X)] = -1

iris_X[:5]
```

```python
## 9. Treat -1 as a missing value and impute it using the mean of each feature.

from sklearn.impute import SimpleImputer

impute = SimpleImputer(
    missing_values=-1,
    strategy="mean"
)

iris_X_prime = impute.fit_transform(iris_X)

iris_X_prime[:5]
```

```python
## 10. Replace missing values with -1 using pandas and NumPy.

import pandas as pd

iris_X_prime = np.where(
    pd.DataFrame(iris_X).isnull(),
    -1,
    iris_X
)

iris_X_prime[:5]
```

```python
## 11. Replace missing values with -1 using the pandas fillna method.

pd.DataFrame(iris_X).fillna(-1)[:5].values
```
